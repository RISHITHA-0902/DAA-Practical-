```python
# Vertex Cover using Greedy Method

n = int(input("Enter number of vertices: "))
e = int(input("Enter number of edges: "))

edges = []

print("Enter edges (u v):")

for i in range(e):
    u, v = map(int, input().split())
    edges.append((u, v))

cover = set()

remaining = edges.copy()

while remaining:

    degree = [0] * n

    # Calculate degree of each vertex
    for u, v in remaining:
        degree[u] += 1
        degree[v] += 1

    # Select vertex with maximum degree
    vertex = degree.index(max(degree))

    cover.add(vertex)

    # Remove edges covered by vertex
    remaining = [
        (u, v) for u, v in remaining
        if u != vertex and v != vertex
    ]

print("Vertex Cover:", sorted(cover))
print("Size of Vertex Cover:", len(cover))
```
