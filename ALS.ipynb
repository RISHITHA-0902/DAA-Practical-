```python
# Assembly Line Scheduling

a = [
    [4, 5, 3, 2],
    [2, 10, 1, 4]
]

t = [
    [0, 7, 4, 5],
    [0, 9, 2, 8]
]

entry = [10, 12]
exit = [18, 7]

n = len(a[0])

T1 = entry[0] + a[0][0]
T2 = entry[1] + a[1][0]

for i in range(1, n):

    newT1 = min(
        T1 + a[0][i],
        T2 + t[1][i] + a[0][i]
    )

    newT2 = min(
        T2 + a[1][i],
        T1 + t[0][i] + a[1][i]
    )

    T1 = newT1
    T2 = newT2

answer = min(T1 + exit[0], T2 + exit[1])

print("Minimum Assembly Line Time:", answer)
```
