```python
# Bin Packing using First Fit Decreasing

items = list(map(int, input("Enter item sizes: ").split()))
capacity = int(input("Enter bin capacity: "))

# Sort items in decreasing order
items.sort(reverse=True)

bins = []

for item in items:

    placed = False

    for i in range(len(bins)):
        if bins[i] + item <= capacity:
            bins[i] += item
            placed = True
            break

    if not placed:
        bins.append(item)

print("Number of bins required:", len(bins))
print("Bins:", bins)
```
