```python
# Set Cover Problem using Greedy Method

universe = set(map(int, input("Enter universe elements: ").split()))

n = int(input("Enter number of sets: "))

sets = []

for i in range(n):
    s = set(map(int, input(f"Enter elements of set {i+1}: ").split()))
    sets.append(s)

covered = set()
selected = []

while covered != universe:

    best_set = None
    best_new = set()

    for i, s in enumerate(sets):
        new_elements = s - covered

        if len(new_elements) > len(best_new):
            best_new = new_elements
            best_set = i

    if best_set is None:
        print("Set cover is not possible")
        break

    selected.append(best_set + 1)
    covered.update(sets[best_set])

print("Selected sets:", selected)
print("Covered elements:", covered)
```
